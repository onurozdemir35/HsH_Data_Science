```python
class  BankAccount:
    def __init__(self, owner, balance):
         self.owner = owner
         self.__balance = balance # Privates Attribut

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            raise ValueError('Amount muss positive sein')

    def get_balance(self):
        return self.__balance


account = BankAccount("Alice", 1000)
account.deposit(1500)
print(account.get_balance()) # Zugriff auf private Daten über Methode

```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    
    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
     def __init__(self,radius):
            self.radius = radius

     def area(self):
            return math.pi * self.radius**2 
     
     def perimeter(self):
            return 2 * math.pi * self.radius

# Example usage:
rect = Rectangle(5, 10)
print(f"Rectangle - Area: {rect.area()}, Perimeter: {rect.perimeter()}")

circle = Circle(7)
print(f"Circle - Area: {circle.area():.2f}, Perimeter: {circle.perimeter():.2f}")
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b
    
    @classmethod
    def pi(cls):
        return 3.14159
    
print(MathUtils.add(5, 7))
print(MathUtils.pi())
```

```python
class TemperaturUmrechner:
    standard_einheit = "Celsius"  # Standard-Einheit für Temperatur

    def __init__(self, celsius):
        self.celsius = celsius

    @staticmethod
    def celsius_zu_fahrenheit(celsius):
        """Wandelt Celsius in Fahrenheit um."""
        return (celsius * 9/5) + 32
    
    @staticmethod
    def fahrenheit_zu_celsius(fahrenheit):
        """Wandelt Fahrenheit in Celsius um."""
        return (fahrenheit - 32) * 5/9
    
    @classmethod
    def von_fahrenheit(cls, fahrenheit):
        """Erstellt eine Temperatur-Instanz aus Fahrenheit."""
        celsius = cls.fahrenheit_zu_celsius(fahrenheit)
        return cls(celsius)
    
    @classmethod
    def set_standart_einheit(cls, einheit):
        """Setzt die Standard-Einheit für Temperatur (Celsius oder Fahrenheit)."""
        if einheit in ["Celsius", "Fahrenheit"]:
            cls.standard_einheit = einheit
        else:
            raise ValueError("Einheit muss 'Celsius' oder 'Fahrenheit' sein.")

# --- Nutzung ---
print(f"Standard-Einheit: {TemperaturUmrechner.standard_einheit}")  # Standard: Celsius

# Ändern der Standard-Einheit auf Fahrenheit
TemperaturUmrechner.set_standart_einheit("Fahrenheit")
print(f"Neue Standard-Einheit: {TemperaturUmrechner.standard_einheit}")  # Jetzt: Fahrenheit
```
